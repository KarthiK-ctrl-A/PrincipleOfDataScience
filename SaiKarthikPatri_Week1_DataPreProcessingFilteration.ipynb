ICP Task 1 -- Sai Karthik Patri

```
df = pd.read_csv('https://raw.githubusercontent.com/datasets/hepatitis/master/data/hepatitis.csv')
```

```
numeric_cols = df.select_dtypes(include='number').columns

print("Missing values in numeric columns before replacement:")
print(df[numeric_cols].isnull().sum())
```

```
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())

print("\nMissing values in numeric columns after replacement:")
print(df[numeric_cols].isnull().sum())

print("\nUpdated dataset (first 10 rows):")
print(df.head(10))
```
